# Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
```

```python
# 1. Create a sample dataset
data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Karan",
             "Neha", "Arjun", "Simran"],
    "Maths": [85, 92, 67, 95, 78, 88, 56, 90],
    "Science": [80, 95, 72, 91, 75, 86, 60, 93],
    "English": [78, 89, 70, 94, 82, 85, 65, 91]
}
df = pd.DataFrame(data)
```

```python
# 2. Calculate average marks
subjects = ["Maths", "Science", "English"]
df["Average"] = df[subjects].mean(axis=1)

# 3. Assign performance categories
df["Result"] = np.where(df["Average"] >= 85, "Excellent",
                 np.where(df["Average"] >= 70, "Good", "Needs Improvement"))

print("Student Performance Dataset:")
print(df)
```

```python
# 4. Overall statistics
print("\nOverall Statistics:")
print(df[subjects + ["Average"]].describe().round(2))
```

```python
# 5. Highest scorer in each subject
print("\nHighest Marks by Subject:")
print(df[subjects].max())
```

```python
# 6. Subject-wise average
print("\nAverage Marks by Subject:")
print(df[subjects].mean().round(2))
```

```python
# 7. Top 3 students by average
print("\nTop 3 Students:")
print(df.nlargest(3, "Average")[["Name", "Average", "Result"]])
```

```python
# 8. Count students in each performance category
print("\nPerformance Category Counts:")
print(df["Result"].value_counts())
```

```python
# 9. Save analysis dataset
df.to_csv("student_performance_eda.csv", index=False)
print("\nEDA dataset saved successfully!")
```
