# Data Cleaning with Pandas

```python
import pandas as pd
import numpy as np
```

```python
# 1. Create a messy dataset
data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Aman", None],
    "Age": [20, 21, np.nan, 22, 20, 19],
    "Marks": [85, 92, 67, np.nan, 85, 74],
    "City": ["Delhi", "Meerut", "Delhi", "Noida", "Delhi", "Meerut"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)
```

```python
# 2. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 3. Fill missing numeric values with column averages
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
# 4. Fill missing names
df["Name"] = df["Name"].fillna("Unknown")
```

```python
# 5. Remove duplicate rows
df = df.drop_duplicates()
```

```python
# 6. Reset row indexes
df = df.reset_index(drop=True)
```

```python
# 7. Display cleaned dataset
print("\nCleaned Dataset:")
print(df)
```

```python
# 8. Verify cleaning
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# 9. Save cleaned data as CSV
df.to_csv("cleaned_students.csv", index=False)

print("\nCleaned dataset saved as cleaned_students.csv")
```
