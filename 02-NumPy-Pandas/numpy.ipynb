# NumPy Basics

```python
import numpy as np
```

```python
# 1. Create an array
marks = np.array([85, 90, 78, 92, 88])

print("Student Marks:", marks)
```

```python
# 2. Basic Statistics
print("Total Marks:", np.sum(marks))
print("Average Marks:", np.mean(marks))
print("Median Marks:", np.median(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
# 3. Array Operations
print("\nMarks + 5:", marks + 5)
print("Marks - 5:", marks - 5)
```

```python
# 4. Filter Values
print("\nMarks above 85:", marks[marks > 85])
```

```python
# 5. Reshape Array
numbers = np.arange(1, 7)
matrix = numbers.reshape(2, 3)

print("\n2D Array:")
print(matrix)
```
