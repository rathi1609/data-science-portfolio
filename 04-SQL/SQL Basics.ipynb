# SQL Basics using SQLite

```python
# 1. Create database connection
connection = sqlite3.connect("student_database.db")
cursor = connection.cursor()

# 2. Create a table
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks REAL,
    city TEXT
)
""")
```

```python
# 3. Insert sample data
cursor.execute("DELETE FROM students")

students = [
    (1, "Aman", 20, 85, "Delhi"),
    (2, "Priya", 21, 92, "Meerut"),
    (3, "Rahul", 19, 67, "Delhi"),
    (4, "Sneha", 22, 95, "Noida"),
    (5, "Karan", 20, 78, "Meerut")
]

cursor.executemany("""
INSERT INTO students
(student_id, name, age, marks, city)
VALUES (?, ?, ?, ?, ?)
""", students)

connection.commit()
```

```python
# 4. SELECT: Display all students
print("All Students:")
df = pd.read_sql_query(
    "SELECT * FROM students", connection
)
print(df)

# 5. WHERE: Students scoring above 80
print("\nStudents scoring above 80:")
df = pd.read_sql_query("""
SELECT name, marks
FROM students
WHERE marks > 80
""", connection)
print(df)
```

```python
# 6. ORDER BY: Sort marks from highest to lowest
print("\nStudents sorted by marks:")
df = pd.read_sql_query("""
SELECT name, marks
FROM students
ORDER BY marks DESC
""", connection)
print(df)
```

```python
# 7. Aggregate functions
print("\nStudent Statistics:")
df = pd.read_sql_query("""
SELECT
    COUNT(*) AS total_students,
    AVG(marks) AS average_marks,
    MAX(marks) AS highest_marks,
    MIN(marks) AS lowest_marks
FROM students
""", connection)
print(df)
```

```python
# 8. GROUP BY: Average marks by city
print("\nAverage Marks by City:")
df = pd.read_sql_query("""
SELECT city, AVG(marks) AS average_marks
FROM students
GROUP BY city
""", connection)
print(df)
```

```python
# 9. Close database connection
connection.close()

print("\nSQL practical completed successfully!")
```

```python
# 6. IN operator
print("\n6. Employees in IT or Finance")
print(pd.read_sql_query("""
    SELECT name, department
    FROM employees
    WHERE department IN ('IT', 'Finance')
""", connection))
```

# SQL Aggregate Functions and GROUP BY

```python
connection = sqlite3.connect(":memory:")

employees = pd.DataFrame({
    "name": ["Aman", "Priya", "Rahul", "Sneha",
             "Karan", "Neha", "Riya"],
    "department": ["IT", "HR", "IT", "Finance",
                   "HR", "IT", "Finance"],
    "salary": [60000, 45000, 75000, 55000,
               40000, 80000, 65000]
})

employees.to_sql("employees", connection, index=False)
```

```python
# 1. COUNT: Total employees
print("1. Total Employees")
print(pd.read_sql_query("""
SELECT COUNT(*) AS total_employees
FROM employees
""", connection))

# 2. SUM: Total salary
print("\n2. Total Salary")
print(pd.read_sql_query("""
SELECT SUM(salary) AS total_salary
FROM employees
""", connection))
```

```python
# 3. AVG: Average salary
print("\n3. Average Salary")
print(pd.read_sql_query("""
SELECT AVG(salary) AS average_salary
FROM employees
""", connection))
```

```python
# 4. MIN and MAX
print("\n4. Lowest and Highest Salary")
print(pd.read_sql_query("""
SELECT
    MIN(salary) AS lowest_salary,
    MAX(salary) AS highest_salary
FROM employees
""", connection))
```

```python
# 5. GROUP BY: Department-wise employee count
print("\n5. Employees by Department")
print(pd.read_sql_query("""
SELECT department, COUNT(*) AS employee_count
FROM employees
GROUP BY department
""", connection))
```

```python
# 6. Department-wise average salary
print("\n6. Average Salary by Department")
print(pd.read_sql_query("""
SELECT department, AVG(salary) AS average_salary
FROM employees
GROUP BY department
""", connection))
```

```python
# 7. HAVING: Departments with average salary above 55000
print("\n7. Departments with Average Salary Above 55000")
print(pd.read_sql_query("""
SELECT department, AVG(salary) AS average_salary
FROM employees
GROUP BY department
HAVING AVG(salary) > 55000
""", connection))

connection.close()
print("\nSQL aggregate practical completed successfully!")
```

# SQL JOINs Practical

```python
connection = sqlite3.connect(":memory:")

# Create Employees table
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5],
    "name": ["Aman", "Priya", "Rahul", "Sneha", "Karan"],
    "department_id": [101, 102, 101, 103, None]
})

# Create Departments table
departments = pd.DataFrame({
    "department_id": [101, 102, 103, 104],
    "department_name": ["IT", "HR", "Finance", "Marketing"]
})

employees.to_sql("employees", connection, index=False)
departments.to_sql("departments", connection, index=False)
```

```python
# 1. Display Employees
print("1. Employees Table")
print(pd.read_sql_query("SELECT * FROM employees", connection))

# 2. Display Departments
print("\n2. Departments Table")
print(pd.read_sql_query("SELECT * FROM departments", connection))
```

```python
# 3. INNER JOIN
print("\n3. INNER JOIN")
print(pd.read_sql_query("""
SELECT e.name, d.department_name
FROM employees e
INNER JOIN departments d
ON e.department_id = d.department_id
""", connection))
```

```python
# 4. LEFT JOIN
print("\n4. LEFT JOIN")
print(pd.read_sql_query("""
SELECT e.name, d.department_name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id
""", connection))
```

```python
# 5. Find employees without a matching department
print("\n5. Employees Without a Department")
print(pd.read_sql_query("""
SELECT e.name
FROM employees e
LEFT JOIN departments d
ON e.department_id = d.department_id
WHERE d.department_id IS NULL
""", connection))
```

```python
# 6. Departments with no employees
print("\n6. Departments Without Employees")
print(pd.read_sql_query("""
SELECT d.department_name
FROM departments d
LEFT JOIN employees e
ON d.department_id = e.department_id
WHERE e.employee_id IS NULL
""", connection))

connection.close()
print("\nSQL JOINs practical completed successfully!")
```

# SQL Subqueries Practical

```python
connection = sqlite3.connect(":memory:")

# Create sample employee data
employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5, 6],
    "name": ["Aman", "Priya", "Rahul", "Sneha", "Karan", "Neha"],
    "department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "salary": [60000, 45000, 75000, 55000, 40000, 70000]
})

employees.to_sql("employees", connection, index=False)
```

```python
# 1. Display all employees
print("1. All Employees")
print(pd.read_sql_query(
    "SELECT * FROM employees", connection
))

# 2. Employees earning above the average salary
print("\n2. Employees Above Average Salary")
print(pd.read_sql_query("""
SELECT name, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
)
""", connection))
```

```python
# 3. Employee with the highest salary
print("\n3. Highest-Paid Employee")
print(pd.read_sql_query("""
SELECT name, salary
FROM employees
WHERE salary = (
    SELECT MAX(salary)
    FROM employees
)
""", connection))
```

```python
# 4. Employees earning more than the average IT salary
print("\n4. Employees Earning More Than Average IT Salary")
print(pd.read_sql_query("""
SELECT name, department, salary
FROM employees
WHERE salary > (
    SELECT AVG(salary)
    FROM employees
    WHERE department = 'IT'
)
""", connection))
```

```python
# 5. Departments with average salary above 55000
print("\n5. Departments Above Average Salary Threshold")
print(pd.read_sql_query("""
SELECT department, AVG(salary) AS average_salary
FROM employees
GROUP BY department
HAVING AVG(salary) > 55000
""", connection))
```

```python
# 6. Employees earning more than at least one HR employee
print("\n6. Employees Earning More Than Some HR Employees")
print(pd.read_sql_query("""
SELECT name, salary
FROM employees
WHERE salary > (
    SELECT MIN(salary)
    FROM employees
    WHERE department = 'HR'
)
ORDER BY salary DESC
""", connection))

connection.close()

print("\nSQL Subqueries practical completed successfully!")
```

# SQL Window Functions

```python
connection = sqlite3.connect(":memory:")

employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5, 6, 7],
    "name": ["Aman", "Priya", "Rahul", "Sneha",
             "Karan", "Neha", "Riya"],
    "department": ["IT", "HR", "IT", "Finance",
                   "HR", "IT", "Finance"],
    "salary": [60000, 45000, 75000, 55000,
               45000, 80000, 65000]
})

employees.to_sql("employees", connection, index=False)
```

```python
# 1. ROW_NUMBER: Unique row number by salary
print("1. ROW_NUMBER")
print(pd.read_sql_query("""
SELECT name, department, salary,
       ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM employees
""", connection))
```

```python
# 2. RANK: Equal salaries get the same rank, with gaps
print("\n2. RANK")
print(pd.read_sql_query("""
SELECT name, salary,
       RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM employees
""", connection))
```

```python
# 3. DENSE_RANK: Equal salaries get the same rank, no gaps
print("\n3. DENSE_RANK")
print(pd.read_sql_query("""
SELECT name, salary,
       DENSE_RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM employees
""", connection))
```

```python
# 4. PARTITION BY: Rank employees within each department
print("\n4. Department-wise Salary Ranking")
print(pd.read_sql_query("""
SELECT name, department, salary,
       DENSE_RANK() OVER (
           PARTITION BY department
           ORDER BY salary DESC
       ) AS department_rank
FROM employees
ORDER BY department, department_rank
""", connection))
```

# SQL CTE and CASE WHEN

```python
connection = sqlite3.connect(":memory:")

employees = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 5, 6, 7],
    "name": ["Aman", "Priya", "Rahul", "Sneha",
             "Karan", "Neha", "Riya"],
    "department": ["IT", "HR", "IT", "Finance",
                   "HR", "IT", "Finance"],
    "salary": [60000, 45000, 75000, 55000,
               40000, 80000, 65000]
})

employees.to_sql("employees", connection, index=False)
```

```python
# 1. CTE: Employees earning above average salary
print("1. Employees Above Average Salary")

query = """
WITH average_salary AS (
    SELECT AVG(salary) AS avg_salary
    FROM employees
)
SELECT name, department, salary
FROM employees
WHERE salary > (SELECT avg_salary FROM average_salary)
ORDER BY salary DESC;
"""

print(pd.read_sql_query(query, connection))
```

```python

# 2. CASE WHEN: Categorize salaries
print("\n2. Salary Categories")

query = """
SELECT name, salary,
    CASE
        WHEN salary >= 70000 THEN 'High'
        WHEN salary >= 50000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category
FROM employees
ORDER BY salary DESC;
"""

print(pd.read_sql_query(query, connection))
```

```python
# 3. Department-wise salary category counts
print("\n3. Salary Categories by Department")

query = """
SELECT department,
    CASE
        WHEN salary >= 70000 THEN 'High'
        WHEN salary >= 50000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_category,
    COUNT(*) AS employee_count
FROM employees
GROUP BY department, salary_category
ORDER BY department, salary_category;
"""

print(pd.read_sql_query(query, connection))
```

```python
# 4. CTE: Department average salary
print("\n4. Department Average Salary")

query = """
WITH department_stats AS (
    SELECT department, AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department
)
SELECT department, ROUND(avg_salary, 2) AS avg_salary
FROM department_stats
WHERE avg_salary > 55000
ORDER BY avg_salary DESC;
"""

print(pd.read_sql_query(query, connection))

connection.close()
print("\nCTE and CASE WHEN practical completed!")
```

# SQL Sales Data Analysis Project

```python
connection = sqlite3.connect(":memory:")
```

```python
# 1. Create sample sales data
sales = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "product": [
        "Laptop", "Phone", "Tablet", "Laptop",
        "Phone", "Headphones", "Tablet", "Phone"
    ],
    "category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Accessories",
        "Electronics", "Electronics"
    ],
    "quantity": [2, 3, 4, 1, 2, 5, 3, 1],
    "price": [50000, 20000, 15000, 50000,
              20000, 2000, 15000, 20000],
    "month": [
        "January", "January", "February", "February",
        "March", "March", "April", "April"
    ]
})

sales.to_sql("sales", connection, index=False)
```

```python
# 2. Display all sales
print("1. All Sales Records")
print(pd.read_sql_query(
    "SELECT * FROM sales", connection
))
```

```python
# 3. Calculate total revenue
print("\n2. Total Revenue")
print(pd.read_sql_query("""
SELECT SUM(quantity * price) AS total_revenue
FROM sales
""", connection))
```

```python
# 4. Revenue by product
print("\n3. Revenue by Product")
print(pd.read_sql_query("""
SELECT product,
       SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC
""", connection))
```

```python
# 5. Revenue by month
print("\n4. Revenue by Month")
print(pd.read_sql_query("""
SELECT month,
       SUM(quantity * price) AS revenue
FROM sales
GROUP BY month
ORDER BY CASE month
    WHEN 'January' THEN 1
    WHEN 'February' THEN 2
    WHEN 'March' THEN 3
    WHEN 'April' THEN 4
    ELSE 5
END
""", connection))
```

```python
# 6. Best-selling products by quantity
print("\n5. Products Ranked by Quantity Sold")
print(pd.read_sql_query("""
SELECT product,
       SUM(quantity) AS total_quantity
FROM sales
GROUP BY product
ORDER BY total_quantity DESC
""", connection))
```
