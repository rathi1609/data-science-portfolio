# Python Loops

```python
# for loop
print("For Loop:")
for i in range(1, 6):
    print(i)
```

```python
# while loop
print("\nWhile Loop:")
count = 1
while count <= 5:
    print(count)
    count += 1
```

```python
# multiplication
print("\nMultiplication Table of 5:")
for i in range(1, 11):
    print(f"5 x {i} = {5 * i}")
```
