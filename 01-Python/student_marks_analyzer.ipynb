# Student Marks Analyzer

```python
students = {
    "Aarav": [85, 90, 78],
    "Priya": [92, 88, 95],
    "Rahul": [55, 60, 58],
    "Kirti": [95, 91, 98]
}

PASS_MARKS = 40
averages = {}
all_marks = []

print("===== STUDENT MARKS ANALYZER =====\n")

for name, marks in students.items():
    average = sum(marks) / len(marks)
    averages[name] = average
    all_marks.extend(marks)

    result = "Pass" if all(mark >= PASS_MARKS for mark in marks) else "Fail"

    print(f"Student: {name}")
    print(f"Marks: {marks}")
    print(f"Average: {average:.2f}")
    print(f"Result: {result}\n")

top_student = max(averages, key=averages.get)

print("===== CLASS SUMMARY =====")
print(f"Class Average: {sum(all_marks) / len(all_marks):.2f}")
print(f"Highest Mark: {max(all_marks)}")
print(f"Lowest Mark: {min(all_marks)}")
print(f"Top Student: {top_student}")
```
